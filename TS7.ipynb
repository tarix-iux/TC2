```python
# Para la siguiente red eléctrica:

# Numeramos los polos de 0 a n=3

#

#     0-------+--Y1----2---Y3--3---

#                      |           /

#                     Y2           / R

#                      |           /

#     1----------------+-------1----

#

from pytc2.general import print_latex, a_equal_b_latex_s, s

from pytc2.cuadripolos import calc_MAI_vtransf_ij_mn

import sympy as sp


L, C= sp.symbols('L C', complex=True)

#      Nodos: 0      1        2        3

Ymai = sp.Matrix([

                 [ 1/(s*L),    0,      0,      -1/(s*L)],

                 [ 0,    1/(s*L),    0,     -1/(s*L)],

                 [ 0,  0,    s*C, -s*C],

                 [ -1/(s*L),    -1/(s*L),      -s*C,      2/(s*L)+s*C ]

                 ])


print_latex( a_equal_b_latex_s('Y_{MAI}', Ymai ))

V2313 = calc_MAI_vtransf_ij_mn(Ymai, ii=2-1, jj=3-1,mm=1-1,nn=3-1)
print_latex( r'T^{{ {:d}{:d} }}_{{ {:d}{:d} }} = '.format(2, 3, 1, 3) +  sp.latex(V2313))

V2131 = calc_MAI_vtransf_ij_mn(Ymai, ii=2-1, jj=1-1,mm=3-1,nn=1-1)
print_latex( r'T^{{ {:d}{:d} }}_{{ {:d}{:d} }} = '.format(2, 1, 3, 1) +  sp.latex(V2131))

V1232 = calc_MAI_vtransf_ij_mn(Ymai, ii=1-1, jj=2-1,mm=3-1,nn=2-1)
print_latex( r'T^{{ {:d}{:d} }}_{{ {:d}{:d} }} = '.format(1, 2, 3, 2) +  sp.latex(V1232))
```
